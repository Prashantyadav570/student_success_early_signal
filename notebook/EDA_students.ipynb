### **Data Cleaning**

```python
df[df["attendance"] > 100]
```

```python
df.loc[df["attendance"] > 100, ["attendance"]] = np.nan   # greaster then 100 attendance marked as null
```

```python
df[df["previous_gpa"] > 10]
```

```python
df.loc[df["previous_gpa"] > 10, ["previous_gpa"]] = np.nan
```

```python
df.isnull().sum()
```

```python
numeric_columns = [
    "attendance",
    "internal_marks",
    "assignment_completion",
    "previous_gpa",
    "failed_subjects",
    "study_hours",
    "previous_semester_marks",
    "practical_marks",
    "class_participation"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
df.isnull().sum()
```

```python
df.describe()
```

### **Visulization**

#### **Risk_Distribution**

```python
df["risk_level"].value_counts()
```

```python
df["risk_level"].value_counts(normalize=True) * 100
```

```python
df["risk_level"].value_counts().plot(kind="bar")

plt.title("Student Risk Distribution")
plt.xlabel("Risk Level")
plt.ylabel("Number of Students")
plt.show()
```

#### **Attendance distribution**

```python
plt.hist(df["attendance"], bins=10)

plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")
plt.show()
```

#### **Internal marks distribution**

```python
plt.hist(df["internal_marks"], bins=10)

plt.title("Internal Marks Distribution")
plt.xlabel("Internal Marks")
plt.ylabel("Number of Students")
plt.show()
```

#### **Risk level vs marks**

```python
df.groupby("risk_level")["internal_marks"].mean()
```

```python
df.groupby("risk_level")["internal_marks"].mean().plot(kind="bar")

plt.title("Average Internal Marks by Risk Level")
plt.xlabel("Risk Level")
plt.ylabel("Average Internal Marks")
plt.show()
```

#### **Correlation**

```python
numeric_df = df.select_dtypes(include="number")
```

```python
correlation = numeric_df.corr()

correlation
```

```python
plt.figure(figsize=(12, 8))

plt.imshow(correlation, cmap="coolwarm")
plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
